```python

import pandas as pd

data = {
    "Product": ["Notebook", "Pen", "Bag", "Notebook",
                "Bottle", "Pen", "Bag", "Notebook"],
    "City": ["Patna", "Delhi", "Patna", "Delhi",
             "Mumbai", "Patna", "Mumbai", "Patna"],
    "Units": [10, 30, 5, 8, 12, 20, 3, 6],
    "Unit_Price": [50, 10, 500, 50, 100, 10, 500, 50]
}

df = pd.DataFrame(data)

print("First 5 rows:")
print(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nDataset information:")
df.info()
```

```python

# Calculate revenue for every sale
df["Revenue"] = df["Units"] * df["Unit_Price"]

# Display the updated dataset
print(df)

```

```python
Total_revenue = df['Revenue'].sum()
print("total revenue",Total_revenue)
```

```python
product_revenue = df.groupby("Product")["Revenue"].sum()

print("Revenue by product:")
print(product_revenue)
```

```python
city_revenue = df.groupby("City")["Revenue"].sum()

print("Revenue by city:")
print(city_revenue)
```

```python
sorted_sales = df.sort_values(
    by="Revenue",
    ascending=False
)

print(sorted_sales)
```

```python
product_summary = df.groupby("Product").agg(
    Total_Units=("Units", "sum"),
    Total_Revenue=("Revenue", "sum")
)

print(product_summary)
```

```python
import matplotlib.pyplot as plt

product_summary["Total_Revenue"].plot(
    kind="bar",
    title="Total Revenue by Product",
    xlabel="Product",
    ylabel="Revenue (₹)"
)


plt.show()
```
